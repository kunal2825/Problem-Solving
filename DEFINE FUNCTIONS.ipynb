```python

#roots of quad equation
import math
def find_roots(a,b,c):
    try:
        d=b**2-4*a*c
        if d<0:
            #this will generate value error
            r1=(-b+math.sqrt(d))/(2*a)
            r2=(-b-math.sqrt(d))/(2*a)
        elif d==0:
            r1=r2=-b/(2*a)
            print("roots are real and equal")
            print("root1: ",r1)
            print("root2: ",r2)
        else:
            r1=(-b+math.sqrt(d))/(2*a)
            r2=(-b-math.sqrt(d))/(2*a)
            print("roots are real and different")
            print("root1: ",r1)
            print("root2: ",r2)
    except ValueError as e:
        print("value error",e)
    except TypeError as e:
        print("type error",e)
        print("coef.must be numbers")
    except NameError as e:
        print("name error",e)
def main():
    try:
        a=float(input("enter a:"))
        b=float(input("enter b:"))
        c=float(input("enter c:"))
        if a==0:
            print("not a quad eqn")
        else:
            find_roots(a,b,c)
    except ValueError as e:
        print("value error",e)
        print("plz enter numerix value")
    except NameError as e:
        print("name error",e)
    except TypeError as e:
        print("type error",e)
main()
```

```python
#fibonacci series
def fibonacci(n):
    if n<=1:
        return 1
    return fibonacci(n-1)+fibonacci(n-2)
n=int(input("enter n: "))
for i in range(n):
    print(fibonacci(i),end=" ")
fibonacci(n)
```
